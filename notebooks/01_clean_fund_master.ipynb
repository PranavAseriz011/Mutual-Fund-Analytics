## Day 2: Fund Master Data Cleaning

### 🎯 Objective
The objective of this notebook is to inspect, clean, and validate the fund master dataset used as a reference for mutual fund analytics.

### 📁 Dataset Information
- **Source:** `data/raw/01_fund_master.csv`
- **Output:** `data/processed/cleaned_fund_master.csv`

### 🛠️ Tools & Technologies
- Python
- Pandas
- Jupyter Notebook

### 🧹 Cleaning Workflow
1. Load and inspect the raw dataset.
2. Review column names and data types.
3. Check missing values and duplicate records.
4. Validate AMFI scheme codes.
5. Inspect important categorical fields.
6. Apply appropriate cleaning rules.
7. Validate and export the cleaned dataset.

### ✅ Expected Outcome
A validated fund master dataset that can serve as a reference for connecting mutual fund information with NAV history, transactions, performance, and other analytical datasets.

```python
import pandas as pd
from pathlib import Path

print("Pandas imported successfully!")
```

```python
# Locate the project root
project_path = Path.cwd().parent

# Define the fund master CSV path
fund_file = project_path / "data" / "raw" / "01_fund_master.csv"

print("Project folder:", project_path)
print("Fund master file exists:", fund_file.exists())
print("File path:", fund_file)
```

```python
# Load the raw fund master dataset
fund_df = pd.read_csv(fund_file)

print("Fund master dataset loaded successfully!")
print("Dataset shape:", fund_df.shape)
```

```python
# Display the first five rows
display(fund_df.head())

# Check column names
print("Column names:")
print(fund_df.columns.tolist())

# Check data types
print("\nData types:")
print(fund_df.dtypes)

# Check missing values
print("\nMissing values:")
print(fund_df.isnull().sum())

# Check duplicate rows
print("\nDuplicate rows:", fund_df.duplicated().sum())
```

```python
# 1. Check duplicate AMFI scheme codes
print(
    "Duplicate AMFI codes:",
    fund_df["amfi_code"].duplicated().sum()
)

# 2. Check invalid AMFI codes
print(
    "Non-positive AMFI codes:",
    (fund_df["amfi_code"] <= 0).sum()
)

# 3. Check blank text values
text_columns = fund_df.select_dtypes(include=["str", "object"]).columns

print("\nBlank text values:")
for column in text_columns:
    blank_count = (
        fund_df[column]
        .fillna("")
        .str.strip()
        .eq("")
        .sum()
    )
    print(f"{column}: {blank_count}")

# 4. Check date conversion
launch_dates = pd.to_datetime(
    fund_df["launch_date"],
    errors="coerce"
)

print("\nInvalid launch dates:", launch_dates.isna().sum())

# 5. Show the complete missing-value summary
print("\nMissing values by column:")
print(fund_df.isnull().sum().to_string())
```

```python
# Check numeric columns for negative values
numeric_columns = [
    "expense_ratio_pct",
    "exit_load_pct",
    "min_sip_amount",
    "min_lumpsum_amount"
]

for column in numeric_columns:
    print(f"\n{column}")
    print("Minimum:", fund_df[column].min())
    print("Maximum:", fund_df[column].max())
    print("Negative values:", (fund_df[column] < 0).sum())

# Inspect category and plan values
print("\nCategories:")
print(fund_df["category"].value_counts().to_string())

print("\nPlans:")
print(fund_df["plan"].value_counts().to_string())

print("\nRisk categories:")
print(fund_df["risk_category"].value_counts().to_string())
```

```python
# Check unique values in important categorical columns
categorical_columns = [
    "category",
    "sub_category",
    "plan",
    "risk_category",
    "sebi_category_code"
]

for column in categorical_columns:
    print(f"\n--- {column} ---")
    print("Unique values:", fund_df[column].nunique())
    print(fund_df[column].value_counts().to_string())

# Check whether percentage values are within 0–100
print("\nExpense ratios outside 0–100:",
      (~fund_df["expense_ratio_pct"].between(0, 100)).sum())

print("Exit loads outside 0–100:",
      (~fund_df["exit_load_pct"].between(0, 100)).sum())

# Check investment amounts equal to or below zero
print("\nNon-positive SIP amounts:",
      (fund_df["min_sip_amount"] <= 0).sum())

print("Non-positive lumpsum amounts:",
      (fund_df["min_lumpsum_amount"] <= 0).sum())
```

```python
# Create a copy of the original dataset
fund_clean = fund_df.copy()

# Convert launch_date into a proper datetime format
fund_clean["launch_date"] = pd.to_datetime(
    fund_clean["launch_date"],
    errors="coerce"
)

# Sort by AMFI scheme code
fund_clean = fund_clean.sort_values(
    by="amfi_code"
).reset_index(drop=True)

# Validate the cleaned dataset
print("Cleaned dataset shape:", fund_clean.shape)
print("Missing values:", fund_clean.isnull().sum().sum())
print("Duplicate rows:", fund_clean.duplicated().sum())
print("Duplicate AMFI codes:", fund_clean["amfi_code"].duplicated().sum())
print("Invalid launch dates:", fund_clean["launch_date"].isnull().sum())
print("Launch date dtype:", fund_clean["launch_date"].dtype)
```

```python
# Define the output path
output_file = project_path / "data" / "processed" / "cleaned_fund_master.csv"

# Create the processed folder if it doesn't exist
output_file.parent.mkdir(parents=True, exist_ok=True)

# Export the cleaned dataset
fund_clean.to_csv(output_file, index=False)

print("Fund Master dataset exported successfully!")
print("Saved location:", output_file)
print("Exported rows:", len(fund_clean))
```
