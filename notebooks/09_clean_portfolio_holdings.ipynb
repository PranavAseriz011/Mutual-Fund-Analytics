# Dataset 9: Portfolio Holdings — Data Cleaning

## Objective

Load, inspect, and clean the mutual fund portfolio holdings dataset while preserving the original data.

## Source File

`data/raw/09_portfolio_holdings.csv`

## Output File

`data/processed/cleaned_portfolio_holdings.csv`

## Workflow

1. Load and inspect the dataset.
2. Check data types and missing values.
3. Identify duplicate records and formatting issues.
4. Validate numeric values and portfolio-related fields.
5. Apply appropriate cleaning steps.
6. Export and verify the cleaned dataset.


```python
# Step 2: Load and inspect portfolio holdings

import pandas as pd
from pathlib import Path

project_path = Path.cwd().parent

input_file = (
    project_path / "data" / "raw" / "09_portfolio_holdings.csv"
)

holdings_df = pd.read_csv(input_file)

print("Dataset loaded successfully!")
print("Dataset shape:", holdings_df.shape)

print("\nColumn names:")
print(holdings_df.columns.tolist())

print("\nData types:")
print(holdings_df.dtypes)

print("\nFirst 5 rows:")
display(holdings_df.head())
```

```python
# Step 3: Initial data quality checks

print("Missing values by column:")
print(holdings_df.isna().sum())

print("\nTotal missing values:", holdings_df.isna().sum().sum())

print("\nDuplicate rows:", holdings_df.duplicated().sum())

print("\nEmpty or whitespace-only text values:")
text_columns = holdings_df.select_dtypes(include="str").columns

for col in text_columns:
    empty_count = holdings_df[col].str.strip().eq("").sum()
    print(f"{col}: {empty_count}")

print("\nFirst 10 rows:")
display(holdings_df.head(10))

print("\nNumeric summary:")
display(holdings_df.describe())
```

```python
# Step 4: Validate numeric values and portfolio dates

numeric_columns = [
    "weight_pct",
    "market_value_cr",
    "current_price_inr",
]

print("Non-positive values:")
for col in numeric_columns:
    print(f"{col}: {(holdings_df[col] <= 0).sum()}")

print("\nPortfolio weight summary:")
display(holdings_df["weight_pct"].describe())

print("\nPortfolio dates:")
print("Invalid dates:", pd.to_datetime(
    holdings_df["portfolio_date"], errors="coerce"
).isna().sum())

print("Earliest date:", holdings_df["portfolio_date"].min())
print("Latest date:", holdings_df["portfolio_date"].max())

print("\nUnique portfolio dates:")
print(holdings_df["portfolio_date"].nunique())

print("\nUnique AMFI codes:", holdings_df["amfi_code"].nunique())
print("Unique stock symbols:", holdings_df["stock_symbol"].nunique())
print("Unique stock names:", holdings_df["stock_name"].nunique())

print("\nSector distribution:")
print(holdings_df["sector"].value_counts())
```

```python
# Step 5: Check portfolio weights by mutual fund

weight_summary = (
    holdings_df.groupby("amfi_code")
    .agg(
        holdings_count=("stock_symbol", "count"),
        total_weight_pct=("weight_pct", "sum"),
        total_market_value_cr=("market_value_cr", "sum"),
    )
    .round(2)
    .sort_values("total_weight_pct", ascending=False)
)

print("Portfolio summary by AMFI code:")
display(weight_summary)

print("\nSummary of total portfolio weights:")
display(weight_summary["total_weight_pct"].describe())

print("\nFunds with total weights above 100%:")
display(weight_summary[weight_summary["total_weight_pct"] > 100])
```

```python
# Step 6: Check duplicate holdings within each fund

holding_key = [
    "amfi_code",
    "stock_symbol",
    "portfolio_date",
]

duplicate_holdings = holdings_df[
    holdings_df.duplicated(subset=holding_key, keep=False)
].sort_values(holding_key)

print("Duplicate fund-stock-date records:", len(duplicate_holdings))

if duplicate_holdings.empty:
    print("No duplicate fund-stock-date records found.")
else:
    display(duplicate_holdings)

print("\nUnique funds:", holdings_df["amfi_code"].nunique())
print("Unique stock symbols:", holdings_df["stock_symbol"].nunique())
print("Total records:", len(holdings_df))
```

```python
# Step 7: Standardize text fields and portfolio dates

# Standardize column names
holdings_df.columns = (
    holdings_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

# Remove leading and trailing whitespace from text fields
text_columns = holdings_df.select_dtypes(include="str").columns

for col in text_columns:
    holdings_df[col] = holdings_df[col].str.strip()

# Convert portfolio_date to datetime
holdings_df["portfolio_date"] = pd.to_datetime(
    holdings_df["portfolio_date"],
    errors="raise"
)

# Verify
print("Dataset shape:", holdings_df.shape)
print("Portfolio date dtype:", holdings_df["portfolio_date"].dtype)
print("Missing values:", holdings_df.isna().sum().sum())
print("Duplicate rows:", holdings_df.duplicated().sum())

print("\nFirst 5 rows:")
display(holdings_df.head())
```

```python
# Step 8: Inspect numeric values and portfolio dates

print("Missing values by column:")
print(holdings_df.isna().sum())

print("\nNumeric summary:")
display(
    holdings_df[
        [
            "weight_pct",
            "market_value_cr",
            "current_price_inr",
        ]
    ].describe().round(2)
)

print("\nLast 5 records:")
display(holdings_df.tail())

print("\nPortfolio dates:")
print(holdings_df["portfolio_date"].value_counts())
```

```python
# Step 9: Export cleaned portfolio holdings

from pathlib import Path

project_path = Path.cwd().parent
processed_folder = project_path / "data" / "processed"
processed_folder.mkdir(parents=True, exist_ok=True)

output_file = (
    processed_folder / "cleaned_portfolio_holdings.csv"
)

holdings_df.to_csv(output_file, index=False)

print("Dataset exported successfully!")
print("Saved to:", output_file)
print("Rows exported:", len(holdings_df))
print("Columns exported:", len(holdings_df.columns))
print("File size (KB):", round(output_file.stat().st_size / 1024, 2))
```

```python
# Step 10: Verify exported portfolio holdings

verified_df = pd.read_csv(output_file)

print("EXPORTED FILE VERIFICATION")
print("-" * 35)

print("Rows:", verified_df.shape[0])
print("Columns:", verified_df.shape[1])
print("Missing values:", verified_df.isna().sum().sum())
print("Duplicate rows:", verified_df.duplicated().sum())

print("\nColumn names:")
print(verified_df.columns.tolist())

print("\nPortfolio dates:")
print(verified_df["portfolio_date"].value_counts())

print("\nFirst 5 rows:")
display(verified_df.head())
```
