# Dataset 7 — Scheme Performance: Data Cleaning

## Objective
Inspect and clean mutual fund scheme performance data to prepare it for reliable analysis.

## Source
`data/raw/07_scheme_performance.csv`

## Output
`data/processed/cleaned_scheme_performance.csv`

## Workflow
1. Load and inspect the dataset.
2. Check data types and missing values.
3. Identify duplicate records.
4. Validate numeric values and performance metrics.
5. Apply appropriate cleaning steps.
6. Validate and export the cleaned dataset.

```python
import pandas as pd
from pathlib import Path

# Define the project directory
project_path = Path.cwd().parent

# Load the raw scheme performance dataset
scheme_file = (
    project_path / "data" / "raw" / "07_scheme_performance.csv"
)

scheme_df = pd.read_csv(scheme_file)

print("Dataset loaded successfully!")
print("Dataset shape:", scheme_df.shape)

print("\nColumn names:")
print(scheme_df.columns.tolist())

print("\nData types:")
print(scheme_df.dtypes)

print("\nFirst five rows:")
display(scheme_df.head())
```

```python
# Check missing values in each column
print("Missing values by column:")
print(scheme_df.isnull().sum())

# Check total missing values
print("\nTotal missing values:", scheme_df.isnull().sum().sum())

# Check duplicate rows
print("\nDuplicate rows:", scheme_df.duplicated().sum())

# Check duplicate AMFI codes
print(
    "Duplicate AMFI codes:",
    scheme_df["amfi_code"].duplicated().sum()
)

# Check unique values in important categorical columns
for col in ["fund_house", "category", "plan", "risk_grade"]:
    print(f"\nUnique values in {col}:")
    print(scheme_df[col].value_counts(dropna=False))

# Display rows with missing values
print("\nRows containing missing values:")
display(scheme_df[scheme_df.isnull().any(axis=1)])
```

```python
# Check for empty or whitespace-only strings
print("Empty strings by column:")

for col in scheme_df.select_dtypes(include=["object", "str"]).columns:
    empty_count = (
        scheme_df[col]
        .astype("string")
        .str.strip()
        .eq("")
        .sum()
    )

    if empty_count > 0:
        print(f"{col}: {empty_count}")

# Inspect the scheme with AMFI code 119598
print("\nInspecting scheme with AMFI code 119598:")

display(
    scheme_df.loc[
        scheme_df["amfi_code"] == 119598
    ].T
)
```

```python
# Review descriptive statistics for numeric columns
print("Numeric column summary:")
display(scheme_df.describe().T)

# Check for infinite values
import numpy as np

numeric_df = scheme_df.select_dtypes(include="number")

print("\nInfinite values by column:")
print(
    pd.Series(
        np.isinf(numeric_df).sum(),
        index=numeric_df.columns
    )
)

# Check important value ranges
print("\nUnique Morningstar ratings:")
print(sorted(scheme_df["morningstar_rating"].unique()))

print("\nUnique risk grades:")
print(sorted(scheme_df["risk_grade"].unique()))

print("\nPlan values:")
print(sorted(scheme_df["plan"].unique()))
```

```python
# Identify schemes with unusually high Sharpe and Sortino ratios
print("Schemes with Sharpe ratio above 2:")

display(
    scheme_df.loc[
        scheme_df["sharpe_ratio"] > 2,
        [
            "amfi_code",
            "scheme_name",
            "category",
            "sharpe_ratio",
            "sortino_ratio",
            "return_3yr_pct",
            "benchmark_3yr_pct",
        ],
    ]
)

print("\nSchemes with Sortino ratio above 3:")

display(
    scheme_df.loc[
        scheme_df["sortino_ratio"] > 3,
        [
            "amfi_code",
            "scheme_name",
            "category",
            "sharpe_ratio",
            "sortino_ratio",
            "return_3yr_pct",
            "benchmark_3yr_pct",
        ],
    ]
)

# Check whether percentage-based metrics have unexpected signs
print("\nNegative expense ratios:", (scheme_df["expense_ratio_pct"] < 0).sum())
print("Negative AUM values:", (scheme_df["aum_crore"] < 0).sum())
print("Non-positive beta values:", (scheme_df["beta"] <= 0).sum())

print(
    "Morningstar ratings outside 1–5:",
    (~scheme_df["morningstar_rating"].between(1, 5)).sum()
)
```

```python
# Check duplicate scheme identifiers
print("Duplicate AMFI codes:", scheme_df["amfi_code"].duplicated().sum())

# Check duplicate scheme names
print("Duplicate scheme names:", scheme_df["scheme_name"].duplicated().sum())

# Check duplicate scheme-name and plan combinations
print(
    "Duplicate scheme-name and plan combinations:",
    scheme_df.duplicated(subset=["scheme_name", "plan"]).sum()
)

# Check data types
print("\nColumn data types:")
print(scheme_df.dtypes)

# Review scheme counts by category and plan
print("\nSchemes by category:")
print(scheme_df["category"].value_counts())

print("\nSchemes by plan:")
print(scheme_df["plan"].value_counts())
```

```python
# Create a separate copy for cleaning
cleaned_scheme_df = scheme_df.copy()

# Remove leading and trailing whitespace from text columns
text_columns = cleaned_scheme_df.select_dtypes(
    include=["object", "str"]
).columns

for col in text_columns:
    cleaned_scheme_df[col] = cleaned_scheme_df[col].str.strip()

# Verify the cleaned dataset
print("Original shape:", scheme_df.shape)
print("Cleaned shape:", cleaned_scheme_df.shape)
print("Missing values:", cleaned_scheme_df.isnull().sum().sum())
print("Duplicate rows:", cleaned_scheme_df.duplicated().sum())

display(cleaned_scheme_df.head())
```

```python
# Inspect the SBI Magnum Gilt Fund row
gilt_row = cleaned_scheme_df.loc[
    cleaned_scheme_df["amfi_code"] == 119120
]

display(gilt_row.T)

print("\nMissing values in this row:")
print(gilt_row.isnull().sum())

print("\nMissing values in the entire cleaned dataset:")
print(cleaned_scheme_df.isnull().sum())
```

```python
# Define the output path
output_file = (
    project_path
    / "data"
    / "processed"
    / "cleaned_scheme_performance.csv"
)

# Export the cleaned dataset
cleaned_scheme_df.to_csv(output_file, index=False)

print("Dataset exported successfully!")
print("Saved to:", output_file)
print("Final shape:", cleaned_scheme_df.shape)
```

```python
# Reload the exported CSV
verified_df = pd.read_csv(output_file)

# Verify the exported dataset
print("Export verification successful!")
print("File exists:", output_file.exists())
print("Exported shape:", verified_df.shape)
print("Missing values:", verified_df.isnull().sum().sum())
print("Duplicate rows:", verified_df.duplicated().sum())

# Confirm the expected columns
print("\nColumn count:", len(verified_df.columns))
print("Columns match:", verified_df.columns.tolist() == cleaned_scheme_df.columns.tolist())
```
