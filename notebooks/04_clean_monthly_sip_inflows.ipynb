## Day 2: Monthly SIP Inflows Data Cleaning

### 🎯 Objective
Inspect, clean, and validate the monthly Systematic Investment Plan (SIP) inflows dataset.

- Pandas

### 🧹 Cleaning Workflow
1. Load and inspect the raw dataset.
2. Review column names and data types.
3. Check missing values and duplicate records.
4. Validate date formats and numeric values.
5. Inspect monthly reporting periods.
6. Apply appropriate cleaning rules.
7. Export and verify the cleaned dataset.

### ✅ Expected Outcome
A validated monthly SIP inflows dataset ready for further analysis in the Mutual Fund Analytics project.

```python
import pandas as pd
from pathlib import Path
```

```python
# Locate the project directory
project_path = Path.cwd().parent

# Define the raw dataset path
sip_file = (
    project_path
    / "data"
    / "raw"
    / "04_monthly_sip_inflows.csv"
)

# Load the raw dataset
sip_df = pd.read_csv(sip_file)

# Inspect the dataset
print("Dataset loaded successfully!")
print("Shape:", sip_df.shape)

print("\nColumn names:")
print(sip_df.columns.tolist())

print("\nData types:")
print(sip_df.dtypes)

display(sip_df.head())
```

```python
# 1. Check missing values
print("MISSING VALUES")
print(sip_df.isnull().sum())

# 2. Check total missing values
print("\nTotal missing values:", sip_df.isnull().sum().sum())

# 3. Check duplicate rows
print("\nComplete duplicate rows:", sip_df.duplicated().sum())

# 4. Check duplicate months
print("Duplicate months:", sip_df["month"].duplicated().sum())

# 5. Validate month format
sip_dates = pd.to_datetime(
    sip_df["month"],
    format="%Y-%m",
    errors="coerce"
)

print("\nInvalid month values:", sip_dates.isna().sum())
print("Earliest month:", sip_dates.min())
print("Latest month:", sip_dates.max())

# 6. Check chronological order
print(
    "Months already sorted:",
    sip_dates.is_monotonic_increasing
)

# 7. Inspect numeric columns
numeric_columns = [
    "sip_inflow_crore",
    "active_sip_accounts_crore",
    "new_sip_accounts_lakh",
    "sip_aum_lakh_crore",
    "yoy_growth_pct"
]

print("\nNUMERIC COLUMN SUMMARY")
display(sip_df[numeric_columns].describe())

# 8. Display rows containing missing values
print("\nROWS CONTAINING MISSING VALUES")
display(sip_df[sip_df.isnull().any(axis=1)])
```

```python
# Check for non-positive values in the main metrics
positive_columns = [
    "sip_inflow_crore",
    "active_sip_accounts_crore",
    "new_sip_accounts_lakh",
    "sip_aum_lakh_crore"
]

print("NON-POSITIVE VALUES")
for column in positive_columns:
    print(f"{column}:", (sip_df[column] <= 0).sum())

# Check year-over-year growth values
print("\nYOY GROWTH SUMMARY")
print(sip_df["yoy_growth_pct"].describe())

print(
    "\nNon-missing YoY growth values outside 0–100%:",
    (
        sip_df["yoy_growth_pct"].notna()
        & ~sip_df["yoy_growth_pct"].between(0, 100)
    ).sum()
)

# Display all monthly records from 2023 onward
print("\nYOY GROWTH VALUES FROM 2023 ONWARD")
display(
    sip_df.loc[
        sip_df["month"] >= "2023-01",
        ["month", "sip_inflow_crore", "yoy_growth_pct"]
    ]
)
```

```python
# Create a copy of the original dataset
sip_clean = sip_df.copy()

# Convert month to datetime format
sip_clean["month"] = pd.to_datetime(
    sip_clean["month"],
    format="%Y-%m",
    errors="coerce"
)

# Sort records chronologically
sip_clean = sip_clean.sort_values(
    by="month"
).reset_index(drop=True)

# Validate the cleaned dataset
print("Cleaned dataset shape:", sip_clean.shape)
print("Missing values:", sip_clean.isnull().sum().sum())
print("Missing YoY growth values:", sip_clean["yoy_growth_pct"].isnull().sum())
print("Duplicate rows:", sip_clean.duplicated().sum())
print("Duplicate months:", sip_clean["month"].duplicated().sum())
print("Invalid months:", sip_clean["month"].isnull().sum())
print("Month dtype:", sip_clean["month"].dtype)

# Verify the reporting period
print("Earliest month:", sip_clean["month"].min())
print("Latest month:", sip_clean["month"].max())
```

```python
# Investigate unusually high new SIP account values
display(
    sip_df.loc[
        sip_df["new_sip_accounts_lakh"] > 15,
        ["month", "new_sip_accounts_lakh"]
    ]
)
```

```python
# Inspect monthly SIP account data around the unusual value
display(
    sip_df.loc[
        sip_df["month"].between("2025-01", "2025-06"),
        ["month", "new_sip_accounts_lakh", "sip_inflow_crore"]
    ]
)
```

```python
# Flag unusually high new SIP account values for review
sip_clean["potential_anomaly"] = (
    sip_clean["new_sip_accounts_lakh"] > 15
)

# Display flagged records
display(
    sip_clean.loc[
        sip_clean["potential_anomaly"],
        ["month", "new_sip_accounts_lakh", "potential_anomaly"]
    ]
)

# Count flagged records
print(
    "Potential anomalies flagged:",
    sip_clean["potential_anomaly"].sum()
)
```

```python
# Define the output path
output_file = (
    project_path
    / "data"
    / "processed"
    / "cleaned_monthly_sip_inflows.csv"
)

# Export the cleaned dataset
sip_clean.to_csv(output_file, index=False)

print("Dataset exported successfully!")
print("Saved to:", output_file)
print("Rows exported:", len(sip_clean))
print("Columns exported:", len(sip_clean.columns))
```

```python
# Reload the exported CSV
sip_check = pd.read_csv(output_file)

# Verify the exported dataset
print("Shape:", sip_check.shape)
print("\nMissing values:")
print(sip_check.isnull().sum())

print("\nDuplicate rows:", sip_check.duplicated().sum())
print("Duplicate months:", sip_check["month"].duplicated().sum())
print(
    "Potential anomalies flagged:",
    sip_check["potential_anomaly"].sum()
)

# Inspect the flagged record
display(
    sip_check.loc[sip_check["potential_anomaly"] == True]
)
```

```python
print(
    sip_check.loc[
        sip_check["potential_anomaly"],
        [
            "month",
            "new_sip_accounts_lakh",
            "potential_anomaly"
        ]
    ].to_string(index=False)
)
```
